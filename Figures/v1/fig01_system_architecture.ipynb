```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch
import numpy as np
import os

# Korean font setup
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# Domain colors
COLORS = {
    'Doc': '#4C72B0', 'Img': '#DD8452', 'Movie': '#55A868',
    'Rec': '#C44E52', 'BGM': '#8172B3'
}

SAVE_DIR = os.path.dirname(os.path.abspath('__file__'))
DPI = 300
```

```python
# fig01_architecture_overview.png - System architecture block diagram

fig, ax = plt.subplots(figsize=(14, 10))
ax.set_xlim(0, 14)
ax.set_ylim(0, 10)
ax.axis('off')

def draw_fancy_box(ax, x, y, w, h, label, color, fontsize=11, text_color='white',
                   alpha=1.0, style='round,pad=0.1', bold=False, sublabel=None):
    box = FancyBboxPatch((x, y), w, h,
                         boxstyle=style,
                         facecolor=color, edgecolor='white',
                         linewidth=1.5, alpha=alpha,
                         zorder=2)
    # Shadow
    shadow = FancyBboxPatch((x + 0.06, y - 0.06), w, h,
                            boxstyle=style,
                            facecolor='#888888', edgecolor='none',
                            linewidth=0, alpha=0.25, zorder=1)
    ax.add_patch(shadow)
    ax.add_patch(box)
    weight = 'bold' if bold else 'normal'
    cy = y + h / 2
    if sublabel:
        ax.text(x + w / 2, cy + 0.12, label, ha='center', va='center',
                fontsize=fontsize, color=text_color, fontweight=weight, zorder=3)
        ax.text(x + w / 2, cy - 0.22, sublabel, ha='center', va='center',
                fontsize=fontsize - 2, color=text_color, alpha=0.85, zorder=3)
    else:
        ax.text(x + w / 2, cy, label, ha='center', va='center',
                fontsize=fontsize, color=text_color, fontweight=weight, zorder=3)

def draw_arrow(ax, x1, y1, x2, y2, color='#444444'):
    ax.annotate('', xy=(x2, y2), xytext=(x1, y1),
                arrowprops=dict(arrowstyle='->', color=color,
                                lw=2.0, connectionstyle='arc3,rad=0.0'),
                zorder=4)

# --- Layer 1: Electron Frontend ---
draw_fancy_box(ax, 3.5, 8.6, 7.0, 0.9,
               'Electron Frontend  (React + TypeScript)',
               '#2E86AB', fontsize=12, bold=True)

# --- Layer 2: Flask Backend API ---
draw_fancy_box(ax, 3.5, 7.3, 7.0, 0.9,
               'Flask Backend API  (Python 3.11)',
               '#457B9D', fontsize=12, bold=True)

# --- Arrow: Frontend -> Backend ---
draw_arrow(ax, 7.0, 8.6, 7.0, 8.2)

# --- Layer 3: Tri-CHEF Engine (large container) ---
engine_box = FancyBboxPatch((1.0, 4.5), 12.0, 2.45,
                             boxstyle='round,pad=0.12',
                             facecolor='#EDF2FB', edgecolor='#3A5A8C',
                             linewidth=2.5, alpha=0.95, zorder=1)
ax.add_patch(engine_box)
ax.text(7.0, 6.72, 'Tri-CHEF Engine', ha='center', va='center',
        fontsize=13, color='#1D3557', fontweight='bold', zorder=3)

# Arrow: Backend -> Engine
draw_arrow(ax, 7.0, 7.3, 7.0, 6.95)

# Sub-boxes inside engine
sub_labels = [
    ('Query Expand\n(KO\u2194EN)', '#5E8FBF'),
    ('Dense 3-Axis\nScoring',      '#4A7C59'),
    ('ASF Filter',                 '#A45C5C'),
    ('Lexical Channel',            '#7B6B9E'),
    ('MPLC Fusion',                '#C47A3A'),
]
sub_w = 2.0
sub_h = 0.82
sub_y = 4.72
gap = (12.0 - len(sub_labels) * sub_w) / (len(sub_labels) + 1)
for i, (lbl, clr) in enumerate(sub_labels):
    sx = 1.0 + gap + i * (sub_w + gap)
    # shadow
    s = FancyBboxPatch((sx + 0.05, sub_y - 0.05), sub_w, sub_h,
                       boxstyle='round,pad=0.08',
                       facecolor='#888', edgecolor='none', alpha=0.2, zorder=1)
    ax.add_patch(s)
    b = FancyBboxPatch((sx, sub_y), sub_w, sub_h,
                       boxstyle='round,pad=0.08',
                       facecolor=clr, edgecolor='white',
                       linewidth=1.2, alpha=0.95, zorder=2)
    ax.add_patch(b)
    ax.text(sx + sub_w / 2, sub_y + sub_h / 2, lbl,
            ha='center', va='center', fontsize=9.5, color='white',
            fontweight='bold', zorder=3, linespacing=1.3)
    # small connecting arrows between sub-boxes
    if i < len(sub_labels) - 1:
        next_sx = 1.0 + gap + (i + 1) * (sub_w + gap)
        ax.annotate('', xy=(next_sx, sub_y + sub_h / 2),
                    xytext=(sx + sub_w, sub_y + sub_h / 2),
                    arrowprops=dict(arrowstyle='->', color='#3A5A8C',
                                   lw=1.5), zorder=4)

# --- Layer 4: Domain databases ---
db_items = ['Doc', 'Img', 'Movie', 'Rec', 'BGM']
db_w = 1.9
db_h = 0.75
db_y = 3.35
db_gap = (12.0 - len(db_items) * db_w) / (len(db_items) + 1)
for i, dom in enumerate(db_items):
    dx = 1.0 + db_gap + i * (db_w + db_gap)
    clr = COLORS[dom]
    # shadow
    s = FancyBboxPatch((dx + 0.05, db_y - 0.05), db_w, db_h,
                       boxstyle='round,pad=0.08',
                       facecolor='#888', edgecolor='none', alpha=0.2, zorder=1)
    ax.add_patch(s)
    b = FancyBboxPatch((dx, db_y), db_w, db_h,
                       boxstyle='round,pad=0.08',
                       facecolor=clr, edgecolor='white',
                       linewidth=1.5, zorder=2)
    ax.add_patch(b)
    db_labels = {
        'Doc':   'Doc DB\n(Milvus)', 'Img':   'Img DB\n(Milvus)',
        'Movie': 'Movie DB\n(Milvus)', 'Rec':   'Rec DB\n(Milvus)',
        'BGM':   'BGM DB\n(Milvus)'
    }
    ax.text(dx + db_w / 2, db_y + db_h / 2, db_labels[dom],
            ha='center', va='center', fontsize=9, color='white',
            fontweight='bold', zorder=3, linespacing=1.3)

# Arrow: Engine -> DB layer
draw_arrow(ax, 7.0, 4.5, 7.0, 4.12)

# --- Title ---
ax.text(7.0, 9.75, 'DB_insight \uc2dc\uc2a4\ud15c \uc544\ud0a4\ud14d\ucc98',
        ha='center', va='center', fontsize=16, color='#1D3557',
        fontweight='bold', zorder=5)

# --- Legend ---
legend_patches = [mpatches.Patch(color=v, label=k) for k, v in COLORS.items()]
ax.legend(handles=legend_patches, loc='lower right',
          fontsize=9, framealpha=0.85, title='Domain', title_fontsize=9)

plt.tight_layout()
out_path = os.path.join(SAVE_DIR, 'fig01_architecture_overview.png')
plt.savefig(out_path, dpi=DPI, bbox_inches='tight')
plt.show()
print(f'Saved: {out_path}')
```

```python
# fig01_pipeline_flow.png - Pipeline flow diagram

fig, ax = plt.subplots(figsize=(16, 6))
ax.set_xlim(0, 16)
ax.set_ylim(0, 6)
ax.axis('off')

# Stage definitions: (label, x_center, is_optional, color)
# Main pipeline y = 3.4
MAIN_Y = 3.4
BOX_W = 1.55
BOX_H = 0.75
SUB_W = 1.4
SUB_H = 0.6

stages = [
    ('Query',           1.1,  False, '#2E86AB'),
    ('Query\nExpand',   2.9,  False, '#5E8FBF'),
    ('Dense Scoring\n(3-axis)', 5.05, False, '#4A7C59'),
    ('ASF',             7.2,  True,  '#A45C5C'),
    ('Lexical',         9.0,  True,  '#7B6B9E'),
    ('Reranker',        10.8, True,  '#C47A3A'),
    ('MPLC\nFusion',   12.6, False, '#457B9D'),
    ('Results',        14.4, False, '#1D3557'),
]

def draw_stage_box(ax, cx, cy, w, h, label, color, optional=False,
                   fontsize=9.5, text_color='white'):
    lw = 1.2
    ec = 'white'
    ls = '-'
    alpha = 1.0
    if optional:
        ec = '#AAAAAA'
        ls = '--'
        alpha = 0.82
    # shadow
    s = FancyBboxPatch((cx - w/2 + 0.05, cy - h/2 - 0.05), w, h,
                       boxstyle='round,pad=0.08',
                       facecolor='#888', edgecolor='none', alpha=0.18, zorder=1)
    ax.add_patch(s)
    b = FancyBboxPatch((cx - w/2, cy - h/2), w, h,
                       boxstyle='round,pad=0.08',
                       facecolor=color, edgecolor=ec,
                       linewidth=lw, linestyle=ls,
                       alpha=alpha, zorder=2)
    ax.add_patch(b)
    ax.text(cx, cy, label, ha='center', va='center',
            fontsize=fontsize, color=text_color,
            fontweight='bold', zorder=3, linespacing=1.3)

# Draw main stage boxes
for (lbl, cx, opt, clr) in stages:
    draw_stage_box(ax, cx, MAIN_Y, BOX_W, BOX_H, lbl, clr, optional=opt)

# Draw arrows between main stages
for i in range(len(stages) - 1):
    cx1 = stages[i][1] + BOX_W / 2
    cx2 = stages[i+1][1] - BOX_W / 2
    ax.annotate('', xy=(cx2, MAIN_Y), xytext=(cx1, MAIN_Y),
                arrowprops=dict(arrowstyle='->', color='#444444', lw=1.8),
                zorder=4)

# Sub-boxes under Dense Scoring
dense_cx = stages[2][1]  # 5.05
sub_items = [
    ('Re: SigLIP2\n(1152d)', '#6DB580'),
    ('Im: BGE-M3\n(1024d)',  '#4A9E6B'),
    ('Z: DINOv2\n(1024d)',   '#2E7D52'),
]
sub_xs = [dense_cx - 1.55, dense_cx, dense_cx + 1.55]
sub_y = 2.05

for sx, (lbl, clr) in zip(sub_xs, sub_items):
    draw_stage_box(ax, sx, sub_y, SUB_W, SUB_H, lbl, clr,
                   optional=False, fontsize=8.5)
    # vertical connector from Dense Scoring box bottom to sub-box
    ax.annotate('', xy=(sx, sub_y + SUB_H / 2),
                xytext=(dense_cx, MAIN_Y - BOX_H / 2),
                arrowprops=dict(arrowstyle='->', color='#4A7C59',
                                lw=1.2,
                                connectionstyle='arc3,rad=0.0'),
                zorder=4)

# Optional stage label
ax.text(9.0, 4.6, '\u2460 Optional stages (dashed border)',
        ha='center', va='center', fontsize=9, color='#888888', style='italic')

# Legend patch for optional
opt_patch = mpatches.Patch(facecolor='#A45C5C', edgecolor='#AAAAAA',
                            linestyle='--', linewidth=1.5,
                            label='Optional stage')
req_patch = mpatches.Patch(facecolor='#4A7C59', edgecolor='white',
                            linewidth=1.5, label='Required stage')
ax.legend(handles=[req_patch, opt_patch], loc='lower right',
          fontsize=9, framealpha=0.85)

# Title
ax.text(8.0, 5.55, 'Tri-CHEF \uac80\uc0c9 \ud30c\uc774\ud504\ub77c\uc778',
        ha='center', va='center', fontsize=15, color='#1D3557',
        fontweight='bold', zorder=5)

plt.tight_layout()
out_path = os.path.join(SAVE_DIR, 'fig01_pipeline_flow.png')
plt.savefig(out_path, dpi=DPI, bbox_inches='tight')
plt.show()
print(f'Saved: {out_path}')
```
